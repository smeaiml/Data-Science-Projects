```python
# Handling Missing data in Pandas.
```

```python
import pandas as pd
```

```python
import numpy as np
```

```python
#This is our small data frame.
df=pd.DataFrame([[5,2.9,8],[1.5,np.NAN,np.NAN],[np.NAN,np.NAN,np.NAN],[np.NAN,6.2,23.]])
```

```python
df
```

```python
# This function replaces all the NaN nul values by assigned integer e.g 0 in our case
df.fillna(0)
```

```python
#Replacing values column values 

df.fillna({0:-5,1:0,2:5})
```

```python
df
```

```python
df.fillna(df.mean())
```

```python
df
```

```python
df.fillna(method='ffill')
```

```python
df
```

```python
df=df.dropna()
```
