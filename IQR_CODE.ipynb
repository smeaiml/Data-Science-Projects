```python
Q1=float(df.height.quantile(0.25))

# we have used the Quantile Function to get the exact Quantile Value
```

```python
Q1
```

```python
Q3=float(df.height.quantile(0.75))
```

```python
Q3
```

```python
IQR=Q3-Q1
```

```python
IQR
```

```python
lower_limit=Q1-1.5*IQR
```

```python
upper_limit=Q3+1.5*IQR
```

```python
lower_limit,upper_limit
```

```python
df[(df.height<lower_limit)|(df.height>upper_limit)]
# These are all the Outliers Present in the Data
```

```python
df[(df.height>lower_limit)&(df.height<upper_limit)]
```
